## Practica #10. Sistema Experto Utilizando Python y Prolog.

### En esta practica implementaremos un sistema experto de animales(animales.pl) usando Python y Prolog. Python se encarga de la interaccion con el usuario y Prolog se encarga del razonamiento mediante hechos y reglas.

## Paso 1: Importamos las librerias necesarias

```python
from pyswip import Prolog
```

## Paso 2: Establecer la conexion con Prolog

```python
prolog = Prolog()

# Cargar el archivo de Prolog
prolog.consult("animales.pl")
```

## Paso 3: Diseñamos una funcion para hacer consultas en Prolog

```python
def consultar_animal(animal):

    print("\n Analizando: ",animal)

    # -----------------------------------------
    # Consultar si es mamifero
    # -----------------------------------------

    resultado = list(
        prolog.query(f"mamifero({animal})")
    )

    if resultado:
        print("Es un mamifero")

    # -----------------------------------------
    # Consultar si es ave
    # -----------------------------------------

    resultado = list(
        prolog.query(f"ave({animal})")
    )

    if resultado:
        print("Es un ave")

    # -----------------------------------------
    # Consultar si es reptil
    # -----------------------------------------

    resultado = list(
        prolog.query(f"reptil({animal})")
    )

    if resultado:
        print("Es un reptil")


    # -----------------------------------------
    # Consultar si es de sangre caliente
    # -----------------------------------------

    resultado = list(
        prolog.query(f"reptil({animal})")
    )

    if resultado:
        print("Es un reptil")


    # -----------------------------------------
    # Consultar si es domestico
    # -----------------------------------------

    resultado = list(
        prolog.query(f"domestico({animal})")
    )

    if resultado:
        print("Es un animal domestico")
```

## Paso 4. Implementamos el Programa Principal

```python
print("======================================")
print("SISTEMA EXPERTO DE ANIMALES")
print("Python + Prolog")
print("======================================")

print("\n Animales disponibles")
print("1.- perro")
print("2.- gato")
print("3.- aguila")
print("4.- serpiente")

animal = input(" \n Escoje un animal: ")

consultar_animal(animal)
```
