## Agregando valores a una lista

### Si queremos agregar nuevos elementos a una lista podemos usar la funcion "append" que nos permitira agregar un elemento al final de la lista.

```python
letras = ['A', 'B', 'C', 'D']
letras
```

```python
letras.append('E')
letras
```

### pero cuando intentamos agregar mas de un elemento vamos a tener un inconveniente ya que las listas no estan obligadas a almacenar un solo tipo de dato, por lo que los nuevos elementos se agregan como una lista.

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
```

```python
letras.append(['F','G'])
letras
```

### Para solventar estos problemas denemos usar la funcion "extend", que nos permitira agregar multiples elementos a una lista.

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
```

```python
letras.extend(['F','G'])
letras
```

## Borrando valores de una lista

### En python "del" es una palabra reservada que nos permite eliminar un objeto y como podemos usar del para eliminar variables tambien podemos usarslo para eliminar elementos de una lista.

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
```

```python
del letras[0]
letras
```

### Tambien se puede eliminar multiples elementos de una lista

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
```

```python
del letras[0:2]
letras
```

### Es muy frecuente que en lugar de eliminar un valor de una coleccion de datos queramos extraer ese valor y manipularlo de alguna manera, para extraer un elemento de una lista podemos usar "pop(...)".

### Nota: por defecto al usar pop extraemos el ultimo elemento.

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
una_letra = letras.pop()
letras
```

### Pero tambien podemos usar "pop(indice)" para extraer el elemento de un indice especifico.

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
una_letra = letras.pop(0)
letras
```

### Es posible que no sepamos el indice del elemento que queramos eliminar pero si el valor, por lo que podemos usar la funcion "index(valor)" para obtener el indice, por ejemplo:

```python
letras = ['A', 'B', 'C', 'D', 'E']
letras
indice_c = letras.index('C')
indice_c
```

```python
una_letras = letras.pop(indice_c)
letras
```

```python
una_letras
```

## Ordenando una lista

### Para ordenar una lista en Python podemos usar el metodo sort(...).

```python
carros = ['bmw', 'audi', 'toyota', 'subaru']
carros
```

```python
carros = ['bmw', 'audi', 'toyota', 'subaru']
carros
carros.sort() #Ascendente
carros
```

```python
carros = ['bmw', 'audi', 'toyota', 'subaru']
carros
carros.sort(reverse = True) #Descendente
carros
```

## Estadisticas Basicas en una Lista

### En Python podemos utilizar una serie de funciones que nos ayudan a conocer mejor una coleccion de datos en la lista.

```python
mi_variable = [0.1761876878188148, 0.4984351384045, 0.0560148901981851, 0.1761876878188148, 0.4984351384045, 0.0560148901981851]
mi_variable
```

```python
min(mi_variable) #Minimo
```

```python
max(mi_variable) #Maximo
```

```python
sum(mi_variable) #Sumatoria
```

### Tambien podemo susar el paquete statistics para calcular valores como la media aritmetica, la mediana, etc.

```python
import statistics as st
```

```python
st.mean(mi_variable) # Media Aritmetica
```

```python
st.median(mi_variable) # Mediana
```

```python
st.pstdev(mi_variable) # Desviacion Estandar Poblacional
```

```python
st.pvariance(mi_variable) # Varianza Poblacional
```
